#### The cost of one ekstra u

A problem of size  N cost N^2 evaluations of u. 

When we add one ekstra N, the expression becomes:
(N+1)(N+1) = N^2+2N+1

If we subtract the original cost (N^2), from the cost when adding one ekstra N we get:

N^2-N^2+2N+1 = 2N+1

This means that each time N is increased by 1, it requires 2N+1 additional evaluations of u.

#### How does increasing N improve u?

An increase in N expands the grid used in the search, which increases the amount of points to choose from. This improves the chance of finding the true optimal point, which is why u grows slightly as N increase:

Complements:
u = 3.4007 (N=50)
u = 3.4015 (N=100)
u = 3.4017 (N=500)
u = 3.4017 (N=1000)

Substitutes:
u = 3.4846 (N=50)
u = 3.4850 (N=100)
u = 3.4851 (N=500)
u = 3.4851 (N=1000)

However,the gain in u shrinks quickly as N increase, while the cost keeps growing, since each additional N cost 2N+1 extra valuations of u. beoynd N=500 the gain start to vanish. This is the case for both the calibrations. A finer grid gives more precise results, but at a certin point the cost of extra evaluations outweighs the improvement in u.

We expect the substitute calibration to be the hardest one to optimize, because it has a bigger flatter area compared to the complement calibration. Even though the substitute calibration reaches the highest utility, it has a large area where the gradient is close to zero. ## 3 Relative Prices and Demand

##### Solving the model for p3, where p3 differs from 0.5 to 3

```python
p3_vec = np.linspace(0.5, 3.0, 30)

results = {}
for sigma_B, label in [(0.40, 'complements'), (3.00, 'substitutes')]:
    model = ConsumerClass(par={'sigma_B': sigma_B})
    s1s, s2s, s3s, x1s, x2s, x3s = [], [], [], [], [], []
    for p3 in p3_vec:
        model.par.p3 = p3
        opt = model.solve(do_print=False)
        x1, x2, x3 = model.quantities(opt.s1, opt.w)
        s1s.append(opt.s1); s2s.append(opt.s2); s3s.append(opt.s3)
        x1s.append(x1); x2s.append(x2); x3s.append(x3)
    results[label] = dict(s1=s1s, s2=s2s, s3=s3s, x1=x1s, x2=x2s, x3=x3s)

fig, axes = plt.subplots(2, 2, figsize=(11,8))
for label in results:
    axes[0,0].plot(p3_vec, results[label]['s2'], label=f'bus ({label})')
    axes[0,1].plot(p3_vec, results[label]['s3'], label=f'train ({label})')
    axes[1,0].plot(p3_vec, results[label]['x2'], label=f'bus ({label})')
    axes[1,1].plot(p3_vec, results[label]['x3'], label=f'train ({label})')
for ax, title in zip(axes.flat, ['bus share','train share','bus quantity','train quantity']):
    ax.set_xlabel('$p_3$'); ax.set_title(title); ax.legend()
fig.tight_layout()
```

#### Discussion

When the bus and train are complements, the consumer can't switch to more bus trips to make up for the increases in train prices. Thus, increasing p3 increases the train's budget share while decreasing the quantity of train trips taken, showing that train demand is relatively inelastic because the quantity decrease doesn't perfectly offset the higher price. The number of bus trips also falls over the same range.

When the bus and train are substitutes, the consumer is able to switch between transportation modes and the train's budget share falls more to compensate for the increase in p3.

Food consumption (x1) barely moves in either calibration, in agreement with the nested structure of the model where price changes in train tickets primarily impacts the allocation of spending within the travel goods nest.

# 5 An extension of the model

#### Removing subsistence-level food consumption from discretionary income

```python
class SubsistenceConsumer(ConsumerClass): #creating a subclass
    def setup(self):
        super().setup() #brings in variables from our_consumer.py
        self.par.x1_min = 3.0  # minimum subsistence quantity of food is 3 units

    def quantities(self, s1, w):
        par = self.par
        I_disc = par.I - par.p1*par.x1_min  # first reserves subsistence food (cost of food * minimum quantity) from I
        assert I_disc > 0, "income too low to cover subsistence food" # if I_disc is negative, raise an error, because the consumer cannot afford the minimum food requirement
        x1 = par.x1_min + s1*I_disc/par.p1 #total food consumption is the minimum subsistence quantity plus any discretionary share of I allocated to food
        x2 = (1-s1)*w*I_disc/par.p2 #total consumption of good 2 (bus trips) is the discretionary share of I allocated to it
        x3 = (1-s1)*(1-w)*I_disc/par.p3 #total consumption of good 3 (train trips) is the discretionary share of I allocated to it
        return x1, x2, x3


    def solve(self, s0=None, do_print=True, **kwargs):
        """ in solve() in our_consumer.py, the reported utility is computed using a hardcoded formula that does not account for the subsistence floor. 
        This override recomputes the utility using the quantities() method, which correctly incorporates the subsistence floor.

        Recompute u here from the same (s1,w) using quantities().
        """
        opt = super().solve(s0=s0, do_print=False, **kwargs)

        x1, x2, x3 = self.quantities(opt.s1, opt.w)
        opt.u = self.utility(x1, x2, x3)

        if do_print:
            print(f"s1={opt.s1:.4f}  w={opt.w:.4f}  u={opt.u:.4f}")

        return opt
```

#### Verification

```python
"""
Verify the model works and that the solution satisfies the budget constraint and the minimum subsistence requirement. 
"""
from types import SimpleNamespace

def check_subsistence_solution(model, do_print=True):
    """ solve, then report quantities, true budget shares, and a budget-balance check """
    opt = model.solve(do_print=False)
    par = model.par

    x1, x2, x3 = model.quantities(opt.s1, opt.w)
    spend = par.p1*x1 + par.p2*x2 + par.p3*x3
    true_share1 = par.p1*x1/par.I
    true_share2 = par.p2*x2/par.I
    true_share3 = par.p3*x3/par.I

    result = SimpleNamespace(
        x1=x1, x2=x2, x3=x3, u=opt.u,
        nested_s1=opt.s1, nested_w=opt.w,
        true_share1=true_share1, true_share2=true_share2, true_share3=true_share3,
        spend=spend, I=par.I,
    )

    if do_print:
        print(f"x1_min = {par.x1_min:.2f}   I = {par.I:.2f}   I_disc = {par.I - par.p1*par.x1_min:.2f}")
        print(f"quantities:      x1={x1:.3f}  x2={x2:.3f}  x3={x3:.3f}")
        print(f"true budget shares (of total I): s1={true_share1:.3f}  s2={true_share2:.3f}  s3={true_share3:.3f}  (sum={true_share1+true_share2+true_share3:.4f})")
        print(f"utility = {opt.u:.4f}")
        print(f"budget check: spend={spend:.6f} vs income={par.I:.6f}  -> {'OK' if abs(spend-par.I)<1e-6 else 'MISMATCH'}")
        print(f"x1 >= x1_min ? {x1:.3f} >= {par.x1_min:.2f} -> {'OK' if x1 >= par.x1_min - 1e-9 else 'VIOLATED'}")

    return result

model = SubsistenceConsumer()
check_subsistence_solution(model)
print()
```

#### Comparison

```python
def utility_loss_pct(cls, I, tax_rate):
    """ % utility lost from a lump-sum tax equal to tax_rate*I, starting from income I """
    model = cls()
    model.par.I = I
    u0 = model.solve(do_print=False).u

    T = tax_rate*I
    model.par.I = I - T
    u1 = model.solve(do_print=False).u

    return (u0 - u1)/u0*100

tax_rate = 0.02   # 2 percent of income, matching Section 4
I_vec = np.linspace(3.2, 15, 30)   # I*(1-tax_rate) must stay above p1*x1_min=3

loss_base = [utility_loss_pct(ConsumerClass, I, tax_rate) for I in I_vec]
loss_sub  = [utility_loss_pct(SubsistenceConsumer, I, tax_rate) for I in I_vec]

fig, ax = plt.subplots(figsize=(7,5))
ax.plot(I_vec, loss_base, label='baseline')
ax.plot(I_vec, loss_sub, label='subsistence extension')
ax.set_xlabel('income $I$ before tax')
ax.set_ylabel('utility lost from a 2% lump-sum tax (%)')
ax.set_title('Welfare cost of a 2% tax, by income level')
ax.legend()
plt.show()
```

#### Summation

For a quick verification of the output, we can examine the set variables x1_min, I, and I_disc, which are correct. The budget shares sum to 1, showing that the entirety of I is being spent, and the budget constraint is followed. Output of x1, food consumption, is above set subsistence level (3 units) in the model extension.

We compare the two models by looking at how much utility is lost from the same lump-sum tax of 2%, at different income levels. At I=10, our baseline calibration, the two models show nearly equivilent utility loss because the subsistence requirement barely binds that far above it. But close to the subsistence floor, the utility in subsistence model is impacted more negatively than the base model. Poorer consumers in the model extension are hit proportionally harder than in the standard model, where every dollar of income is treated the same regardless of who's spending it. The extension model asymptotically approaches the utility of the base model where the subsistence requirement becomes a smaller and smaller factor in utility as its share of income becomes infinitely smaller.
